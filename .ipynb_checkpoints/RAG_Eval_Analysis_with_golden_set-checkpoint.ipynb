```python
# Import & Paths

# --- Cell 1: Imports & paths ---
from pathlib import Path
import json, math, re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

# Edit these if files are elsewhere
RESULTS_PATH = Path("rag_eval_results.jsonl")
GOLD_PATH    = Path("iycf_eval_questions_gold_single_kb.json")

assert RESULTS_PATH.exists(), f"Missing: {RESULTS_PATH}"
assert GOLD_PATH.exists(), f"Missing: {GOLD_PATH}"
```

```python
# --- Cell 2: Load model results & gold, then merge on (id, category) ---

# Load results JSONL (skip final summary row)
records = []
with RESULTS_PATH.open("r", encoding="utf-8") as f:
    for line in f:
        line = line.strip()
        if not line:
            continue
        obj = json.loads(line)
        if "summary" in obj:
            continue
        records.append(obj)
df_model = pd.json_normalize(records)

# Load gold
df_gold = pd.read_json(GOLD_PATH)

# Merge
df = df_model.merge(
    df_gold[["id","category","expected_in_kb","gold_answer"]],
    on=["id","category"], how="left"
)

print("Rows:", len(df))
display(df.head(3))
```

```python
# metric heper

# --- Cell 3: Metric helpers (BLEU-4, ROUGE-1/2 F1, ROUGE-L recall, Jaccard) ---

def tokenize(s: str):
    return [t for t in (s or "").lower().strip().split() if t]

def ngrams(tokens, n):
    return list(zip(*[tokens[i:] for i in range(n)])) if len(tokens) >= n else []

def modified_precision(hyp, ref, n):
    hyp_ngrams = ngrams(hyp, n)
    ref_ngrams = ngrams(ref, n)
    if not hyp_ngrams:
        return 0.0
    ref_counts = {}
    for ng in ref_ngrams:
        ref_counts[ng] = ref_counts.get(ng, 0) + 1
    hyp_counts = {}
    for ng in hyp_ngrams:
        hyp_counts[ng] = hyp_counts.get(ng, 0) + 1
    clipped = sum(min(c, ref_counts.get(ng, 0)) for ng, c in hyp_counts.items())
    total = sum(hyp_counts.values())
    return clipped / total if total > 0 else 0.0

def bleu4(hyp_tokens, ref_tokens, smooth=True):
    p_ns = []
    for n in range(1, 5):
        p = modified_precision(hyp_tokens, ref_tokens, n)
        if smooth and p == 0:
            p = 1e-9   # epsilon smoothing
        p_ns.append(p)
    geo_mean = math.exp(sum(math.log(p) for p in p_ns) / 4.0)
    r, c = len(ref_tokens), len(hyp_tokens)
    if c == 0:
        return 0.0
    bp = 1.0 if c > r else math.exp(1 - r / c)
    return bp * geo_mean

def rouge_f1(hyp_tokens, ref_tokens, n):
    hyp_set = ngrams(hyp_tokens, n)
    ref_set = ngrams(ref_tokens, n)
    if not hyp_set and not ref_set:
        return 1.0
    hyp_counts = {}
    for ng in hyp_set:
        hyp_counts[ng] = hyp_counts.get(ng, 0) + 1
    ref_counts = {}
    for ng in ref_set:
        ref_counts[ng] = ref_counts.get(ng, 0) + 1
    overlap = sum(min(c, ref_counts.get(ng, 0)) for ng, c in hyp_counts.items())
    prec = overlap / max(1, len(hyp_set))
    rec  = overlap / max(1, len(ref_set))
    return 0.0 if (prec + rec) == 0 else 2 * prec * rec / (prec + rec)

def lcs_len(a, b):
    m, n = len(a), len(b)
    dp = [[0]*(n+1) for _ in range(m+1)]
    for i in range(m):
        ai = a[i]
        for j in range(n):
            dp[i+1][j+1] = dp[i][j] + 1 if ai == b[j] else max(dp[i][j+1], dp[i+1][j])
    return dp[m][n]

def rouge_l_recall(hyp_tokens, ref_tokens):
    return (lcs_len(hyp_tokens, ref_tokens) / len(ref_tokens)) if ref_tokens else 0.0

def jaccard_unigram(hyp_tokens, ref_tokens):
    hs, rs = set(hyp_tokens), set(ref_tokens)
    if not hs and not rs:
        return 1.0
    return len(hs & rs) / max(1, len(hs | rs))
```

```python
# --- Cell 4: Compute per-record similarity & abstention metrics ---

ABSTAIN_REGEX = re.compile(
    r"(not\s+in\s+(?:the\s+)?(?:knowledge\s*base|kb|context)|"
    r"insufficient\s+information|"
    r"cannot\s+answer|unable\s+to\s+answer|"
    r"no\s+(?:info|information)\s+available|"
    r"unknown|not\s+found|out\s+of\s+scope)",
    flags=re.IGNORECASE
)
# Conservative context-grounding fallback thresholds
NOVELTY_THR = 0.55
ROUGE_THR   = 0.15
CTX_SIM_THR = 0.15

rows = []
for _, r in df.iterrows():
    qid = r["id"]; cat = r["category"]
    ans  = (r.get("answer") or "")
    gold = (r.get("gold_answer") or "")
    in_kb = bool(r.get("expected_in_kb")) if "expected_in_kb" in r else None

    # Tokenize
    ht, rt = tokenize(ans), tokenize(gold)

    # Similarity metrics only for in-KB items
    if in_kb:
        m_bleu  = bleu4(ht, rt)
        m_r1    = rouge_f1(ht, rt, 1)
        m_r2    = rouge_f1(ht, rt, 2)
        m_rl    = rouge_l_recall(ht, rt)
        m_jac   = jaccard_unigram(ht, rt)
        exact   = int(ans.strip().lower() == gold.strip().lower())
    else:
        m_bleu = m_r1 = m_r2 = m_rl = m_jac = np.nan
        exact = np.nan

    # Abstention predictions
    pred_abstain_text = bool(ABSTAIN_REGEX.search(ans))
    a_ctx     = r.get("metrics.a_ctx_cosine", np.nan)
    rouge_ctx = r.get("metrics.rougeL_recall_a_ctx", np.nan)
    novelty   = r.get("metrics.answer_novelty_vs_ctx", np.nan)
    pred_abstain_metric = (
        pd.notnull(a_ctx) and pd.notnull(rouge_ctx) and pd.notnull(novelty) and
        (a_ctx <= CTX_SIM_THR) and (rouge_ctx <= ROUGE_THR) and (novelty >= NOVELTY_THR)
    )
    pred_abstain = pred_abstain_text or pred_abstain_metric
    expected_abstain = not in_kb if in_kb is not None else None

    rows.append({
        "id": qid, "category": cat,
        "answer": ans, "gold_answer": gold,
        "expected_in_kb": in_kb, "expected_abstain": expected_abstain,
        "pred_abstain_text": pred_abstain_text, "pred_abstain_metric": pred_abstain_metric,
        "pred_abstain": pred_abstain,
        "bleu4": m_bleu, "rouge1_f1": m_r1, "rouge2_f1": m_r2, "rougeL_recall": m_rl, "jaccard_uni": m_jac,
        "exact_match": exact,
        "response_length": r.get("response_length", np.nan), "within_char_cap": r.get("within_char_cap", np.nan)
    })

df_eval = pd.DataFrame(rows)
display(df_eval.head(5))
```

```python
# --- Cell 5: Overall abstention confusion matrix & metrics (positive = abstain) ---

tp = int(((df_eval["pred_abstain"] == True)  & (df_eval["expected_abstain"] == True)).sum())
fp = int(((df_eval["pred_abstain"] == True)  & (df_eval["expected_abstain"] == False)).sum())
fn = int(((df_eval["pred_abstain"] == False) & (df_eval["expected_abstain"] == True)).sum())
tn = int(((df_eval["pred_abstain"] == False) & (df_eval["expected_abstain"] == False)).sum())

precision = tp / (tp + fp) if (tp + fp) > 0 else np.nan
recall    = tp / (tp + fn) if (tp + fn) > 0 else np.nan
f1        = (2 * precision * recall / (precision + recall)) if (precision > 0 and recall > 0) else np.nan
accuracy  = (tp + tn) / max(1, (tp + fp + fn + tn))

abstention_overall = pd.DataFrame([{
    "TP_correct_abstain": tp, "FP_over_abstain": fp,
    "FN_missed_abstain": fn, "TN_correct_answer": tn,
    "precision_abstain": round(precision, 4) if precision==precision else None,
    "recall_abstain":    round(recall,    4) if recall==recall else None,
    "f1_abstain":        round(f1,        4) if f1==f1 else None,
    "accuracy":          round(accuracy,  4),
}])

display(abstention_overall)
```

```python
# --- Cell 6: Per-category medians (in-KB only) + abstention metrics ---

# In-KB similarity medians
in_kb_df = df_eval[df_eval["expected_in_kb"] == True].copy()
sim_cols = ["bleu4","rouge1_f1","rouge2_f1","rougeL_recall","jaccard_uni","response_length","exact_match"]
cat_medians = (
    in_kb_df.groupby("category")[sim_cols]
            .median()
            .round(4)
            .reset_index()
            .rename(columns={"exact_match":"exact_match_median"})  # for clarity
)

# Per-category abstention metrics
def per_cat_conf(g):
    tp = int(((g["pred_abstain"] == True)  & (g["expected_abstain"] == True)).sum())
    fp = int(((g["pred_abstain"] == True)  & (g["expected_abstain"] == False)).sum())
    fn = int(((g["pred_abstain"] == False) & (g["expected_abstain"] == True)).sum())
    tn = int(((g["pred_abstain"] == False) & (g["expected_abstain"] == False)).sum())
    prec = tp / (tp + fp) if (tp + fp) > 0 else np.nan
    rec  = tp / (tp + fn) if (tp + fn) > 0 else np.nan
    f1   = (2*prec*rec/(prec+rec)) if (prec>0 and rec>0) else np.nan
    return pd.Series({
        "n": len(g),
        "expected_abstain": int(g["expected_abstain"].sum()),
        "pred_abstain": int(g["pred_abstain"].sum()),
        "abstain_precision": round(prec, 4) if prec==prec else np.nan,
        "abstain_recall":    round(rec,  4) if rec==rec else np.nan,
        "abstain_f1":        round(f1,   4) if f1==f1 else np.nan,
        "TP": tp, "FP": fp, "FN": fn, "TN": tn,
    })

cat_abstain = df_eval.groupby("category").apply(per_cat_conf).reset_index()

# Merge
cat_summary = (
    cat_abstain.merge(cat_medians, on="category", how="left")
               .sort_values("category")
)
display(cat_summary)

# Optional: save
cat_summary.to_csv("gold_eval_category_summary.csv", index=False)
df_eval.to_csv("gold_eval_detailed.csv", index=False)
print("Saved gold_eval_category_summary.csv and gold_eval_detailed.csv")
```

```python
# --- Cell 7: Grouped bar charts (Mean vs Median) for similarity metrics (in-KB only) ---

# Compute means and medians per category for similarity metrics
in_kb_df = df_eval[df_eval["expected_in_kb"] == True].copy()
METRICS = ["bleu4","rouge1_f1","rouge2_f1","rougeL_recall","jaccard_uni","response_length"]

means   = in_kb_df.groupby("category")[METRICS].mean().reset_index()
medians = in_kb_df.groupby("category")[METRICS].median().reset_index()

cats = sorted(in_kb_df["category"].unique())
means  = means.set_index("category").reindex(cats).reset_index()
medians= medians.set_index("category").reindex(cats).reset_index()

from numpy import arange

for m in METRICS:
    fig, ax = plt.subplots()
    x = arange(len(cats))
    width = 0.35
    ax.bar(x - width/2, means[m].fillna(0).values,   width=width, label="Mean")
    ax.bar(x + width/2, medians[m].fillna(0).values, width=width, label="Median")
    ax.set_title(f"Mean vs Median {m} by category (in-KB only)")
    ax.set_ylabel(m)
    ax.set_xticks(x)
    ax.set_xticklabels(cats, rotation=30, ha="right")
    ax.legend()
    plt.tight_layout()
    plt.show()
```

```python
# --- Cell 8: Grouped bars for abstention precision vs recall by category ---

bars = cat_summary[["category","abstain_precision","abstain_recall"]].copy()
bars = bars.fillna(0.0)
cats = list(bars["category"])
x = np.arange(len(cats))
width = 0.35

fig, ax = plt.subplots()
ax.bar(x - width/2, bars["abstain_precision"].values, width=width, label="Precision (abstain)")
ax.bar(x + width/2, bars["abstain_recall"].values,    width=width, label="Recall (abstain)")
ax.set_title("Abstention Precision vs Recall by Category")
ax.set_ylabel("Score")
ax.set_xticks(x)
ax.set_xticklabels(cats, rotation=30, ha="right")
ax.legend()
plt.tight_layout()
plt.show()
```

```python
# --- Cell 9: Confusion matrices (per category) ---

def cat_confusion_table(df_eval):
    rows = []
    for cat, g in df_eval.groupby("category"):
        tp = int(((g["pred_abstain"] == True)  & (g["expected_abstain"] == True)).sum())
        fp = int(((g["pred_abstain"] == True)  & (g["expected_abstain"] == False)).sum())
        fn = int(((g["pred_abstain"] == False) & (g["expected_abstain"] == True)).sum())
        tn = int(((g["pred_abstain"] == False) & (g["expected_abstain"] == False)).sum())
        rows.append({"category": cat, "TP": tp, "FP": fp, "FN": fn, "TN": tn})
    return pd.DataFrame(rows).sort_values("category")

conf_table = cat_confusion_table(df_eval)
display(conf_table)

# Optional: neat print
print(conf_table.to_string(index=False))
```
